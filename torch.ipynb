```python
import torch

x = torch.tensor(3., requires_grad=True) # Initialize a tensor with gradient tracking enabled
y = x ** 2

y.backward() # Compute the gradient of y with respect to x
print(x)       # Output: tensor(3., requires_grad=True)
print(x.grad)  # Output: tensor(6.) -> dy/dx = 2 * x = 2 * 3 = 6
print(y)       # Output: tensor(9., grad_fn=<PowBackward0>) -> y = x^2 = 3^2 = 9
```

```python
import torch.nn as nn

horas = torch.tensor([[1.], [2.], [3.], [4.], [5.], [6.]])  # Input tensor representing hours studied
notas = torch.tensor([[2.5], [4.], [5.5], [7.], [8.], [10.]])  # Target tensor representing grades
```

```python
modelo = nn.Linear(in_features=1, out_features=1)  # Create a linear regression model with 1 input and 1 output feature

print(modelo)
print(list(modelo.parameters()))  # Output: [Parameter containing: tensor([[0.]]), Parameter containing: tensor([0.])] -> The model has a linear layer with bias
```

```python
loss_fn = nn.MSELoss()  # Define the Mean Squared Error loss function

otimizador = torch.optim.SGD(modelo.parameters(), lr=0.01)  # Define the optimizer (Stochastic Gradient Descent) with learning rate 0.01
```

```python
print(horas)
print(modelo(horas))  # Output: tensor([[0.], [0.], [0.], [0.], [0.], [0.]], grad_fn=<AddmmBackward0>) -> The model's predictions for the input hours

#Resultado de -> peso @ horas + viés -> weight @ horas + bias
# W e B são inicialmente aleatórios, criados pelo modelo inicial 
```

```python
for epoca in range(201):
    previsao = modelo(horas) #foward
    loss = loss_fn(previsao, notas) #calculo do erro

    otimizador.zero_grad() #zera o gradiente
    loss.backward()        #backward: calcula gradientes
    otimizador.step()      #ajusta peso e viés

    if epoca % 20 == 0:
        print(f"Época {epoca}: loss: {loss.item():.4f}")
        previsoes = modelo(horas)
        for prev in previsao:
            print(f"Previsão: {prev.item():.4f}", end=" | ")

        print()
        # for h, real, prev in zip(horas, notas, previsoes):
        #     print(f"Horas: {h.item():.1f} | Nota real: {real.item():.1f} | Previsto: {prev.item():.2f}")
```
